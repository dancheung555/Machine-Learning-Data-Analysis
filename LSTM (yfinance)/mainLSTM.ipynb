# LSTM model using stock market prices

In this notebook, I will implement an LSTM model and train it based on stock market prices using yahoo finance (yfinance library).

```python
import yfinance as yf
import matplotlib.pyplot as plt

data = yf.download('HSI', start='2016-01-01', end='2022-12-31')

data.head(10)
```

```python
plt.figure(figsize=(16,8))
plt.plot(data['Close'], color='black')
plt.xlabel('Date')
plt.xticks(rotation=45)
plt.ylabel('Price ($)')
plt.title('HSI Stock Price (5-years)')
plt.grid(True)
plt.show()
```
